```python
import geopandas as gpd
import pandas as pd
from pathlib import Path
```

```python
## Set up file paths
## project root is two levels above notebooks/raw_data_processing
PROJECT_ROOT = Path.cwd().parent.parent
SHAPEFILE = PROJECT_ROOT / 'data' / 'raw' / 'fsa_boundary' / 'gfsa000a11a_e.shp'
SAVE_PATH = PROJECT_ROOT / 'data' / 'processed' / 'fsa_centroids.csv'
```

```python
# Load the FSA shapefile
fsa_gdf = gpd.read_file(SHAPEFILE)

print("Column names:")
print(fsa_gdf.columns.tolist())

print("\nDataFrame info:")
print(fsa_gdf.info())

print("\nFirst few rows:")
print(fsa_gdf.head())

print("\nData types:")
print(fsa_gdf.dtypes)

print(f"\nShape: {fsa_gdf.shape} (rows, columns)")

print("\nSample row (first FSA):")
print(fsa_gdf.iloc[0])
```

```python
print("\nShapefile CRS:")
print(fsa_gdf.crs)
```

```python
fsa_gdf['centroid'] = fsa_gdf.geometry.centroid

# Extract latitude and longitude from centroids
fsa_gdf['longitude'] = fsa_gdf['centroid'].x
fsa_gdf['latitude'] = fsa_gdf['centroid'].y
fsa_gdf = fsa_gdf.rename(columns={'CFSAUID': 'FSA'})
fsa_gdf[['FSA', 'longitude', 'latitude']].to_csv(SAVE_PATH, index=False)
```
