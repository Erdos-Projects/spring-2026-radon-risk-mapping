```python
import pandas as pd
import numpy as np
import geopandas as gpd
import sys
from pathlib import Path
from ydata_profiling import ProfileReport


PROJECT_ROOT = Path.cwd().parent.parent
sys.path.append(str(PROJECT_ROOT))

# Get file location stuff
from src.data.data_loading import load_full_dataset
```

```python
# Define the project root
report_path = PROJECT_ROOT / "figures" / "eda" / "radon_dataset_ydata_profile.html"
report_path.parent.mkdir(parents=True, exist_ok=True)
```

```python
# load the data

df = load_full_dataset()
df = df.copy()
# remove held-out test data
# df = df[~df["is_test"]].copy()
```

```python
# look at what is here
df.head()
df.columns
```

```python
# Get rid of things

# get rid of stuff we don't want to use as features
columns_to_drop = [
    "FSA",
    "longitude",
    "latitude",
    "geometry",
    "spatial_cluster",
    "is_test",
    "cv_fold",
    "provinceterritory"
]

# get rid of bogus features
cols_all_zero = [
    "geolprov_arctic_continental_shelf",
    "geolprov_arctic_platform",
    "geolprov_bear_province",
    "geolprov_innuitian_orogen",
    "geolprov_slave_province",
    "geolprov_oceanic_crust",
    "sedi_weathered_bedrock_or_regolith"
]

df_profile = df.drop(columns=columns_to_drop)
df_profile = df_profile.drop(columns=cols_all_zero)
```

```python
# Add log of Radon concentraiton
df_profile["log_concentration"] = np.log(df_profile["concentration"])
```

```python
# call yData Profiling
profile = ProfileReport(
    df_profile,
    title="Radon FSA Dataset Profiling",
    explorative=True
)

profile.to_file(report_path)
```
