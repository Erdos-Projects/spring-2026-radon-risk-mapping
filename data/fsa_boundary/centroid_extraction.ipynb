```python
import geopandas as gpd
import pandas as pd
from pathlib import Path
```

```python
# Load the FSA shapefile
fsa_gdf = gpd.read_file('gfsa000a11a_e.shp')

print("Column names:")
print(fsa_gdf.columns.tolist())

print("\nDataFrame info:")
print(fsa_gdf.info())

print("\nFirst few rows:")
print(fsa_gdf.head())

print("\nData types:")
print(fsa_gdf.dtypes)

print(f"\nShape: {fsa_gdf.shape} (rows, columns)")

print("\nSample row (first FSA):")
print(fsa_gdf.iloc[0])
```

```python
print("\nShapefile CRS:")
print(fsa_gdf.crs)
fsa_gdf.to_crs('EPSG:4326', inplace=True)
print("\nConverted CRS:")
print(fsa_gdf.crs)
```

```python
fsa_gdf['centroid'] = fsa_gdf.geometry.centroid

# Extract latitude and longitude from centroids
fsa_gdf['longitude'] = fsa_gdf['centroid'].x
fsa_gdf['latitude'] = fsa_gdf['centroid'].y

fsa_gdf[['CFSAUID', 'longitude', 'latitude']].to_csv('fsa_centroids.csv', index=False)
```

```python
print("\nDataFrame info:")
print(fsa_gdf.info())
```

```python
import folium
from folium import GeoJson

# Create base map centered on Canada
m = folium.Map(location=[56, -106], zoom_start=4)

# Add FSA boundaries with popup showing FSA code and other info
for idx, row in fsa_gdf.iterrows():
    # Create popup text with desired information
    popup_text = f"""<b>FSA Code:</b> {row['CFSAUID']}<br>
                     <b>Name:</b> {row['PRNAME']} <br>
                     <b>Centroid coordinate (long,lat):</b> {(row['longitude'], row['latitude'])}"""
    
    folium.GeoJson(
        row['geometry'],
        popup=folium.Popup(popup_text, max_width=300),
        style_function=lambda x: {'color': 'blue', 'fillOpacity': 0.1}
    ).add_to(m)

m.save('canada_fsa_map.html')  # Opens in browser
```
